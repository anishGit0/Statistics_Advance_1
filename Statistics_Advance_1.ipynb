```python
# 3.
def bernoulli_pdf(x, p):
    """
    Compute the probability distribution function (PDF) of a Bernoulli distribution.

    Parameters:
    x (int): The outcome (0 or 1).
    p (float): The probability of success (0 <= p <= 1).

    Returns:
    float: The probability P(X = x).
    """
    if x not in [0, 1]:
        raise ValueError("x must be 0 or 1")
    if not (0 <= p <= 1):
        raise ValueError("p must be between 0 and 1")

    return (p ** x) * ((1 - p) ** (1 - x))

# Example usage:
print(bernoulli_pdf(0, 0.7))  # Expected output: 0.3
print(bernoulli_pdf(1, 0.7))  # Expected output: 0.7
```

```python
# 19.
import numpy as np
import scipy.stats as stats

def z_test(sample, pop_mean, pop_std, alpha=0.05):
    """Performs a one-sample Z-test."""
    sample_mean = np.mean(sample)
    sample_size = len(sample)
    standard_error = pop_std / np.sqrt(sample_size)

    # Calculate Z-score
    z_score = (sample_mean - pop_mean) / standard_error

    # Calculate p-value (two-tailed test)
    p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))

    # Hypothesis Decision
    if p_value < alpha:
        result = "Reject the null hypothesis (H0). Significant difference exists."
    else:
        result = "Fail to reject the null hypothesis (H0). No significant difference."

    return z_score, p_value, result

# Example Data
sample_data = [55, 58, 60, 63, 62, 65, 67, 70, 68, 66]  # Sample dataset
population_mean = 60  # Hypothetical population mean
population_std = 5    # Population standard deviation

# Perform Z-test
z_stat, p_val, decision = z_test(sample_data, population_mean, population_std)

# Output results
print(f"Z-Score: {z_stat:.2f}")
print(f"P-Value: {p_val:.4f}")
print(f"Decision: {decision}")
```

```python
# 20.
import numpy as np
import scipy.stats as stats

def confidence_interval(data, confidence=0.95):
    """Computes the confidence interval for a dataset."""
    sample_mean = np.mean(data)
    sample_std = np.std(data, ddof=1)  # Sample standard deviation (unbiased)
    sample_size = len(data)

    # Get Z-score for the given confidence level
    z_value = stats.norm.ppf((1 + confidence) / 2)

    # Compute Margin of Error (ME)
    margin_of_error = z_value * (sample_std / np.sqrt(sample_size))

    # Compute confidence interval
    lower_bound = sample_mean - margin_of_error
    upper_bound = sample_mean + margin_of_error

    return lower_bound, upper_bound

# Example Dataset
sample_data = [55, 58, 60, 63, 62, 65, 67, 70, 68, 66]  # Sample observations

# Calculate 95% Confidence Interval
ci_lower, ci_upper = confidence_interval(sample_data)

# Output result
print(f"95% Confidence Interval: ({ci_lower:.2f}, {ci_upper:.2f})")
```

```python
# 21.
import numpy as np
import scipy.stats as stats

def confidence_interval(data, confidence=0.95):
    """Computes the confidence interval for a dataset."""
    sample_mean = np.mean(data)
    sample_std = np.std(data, ddof=1)  # Sample standard deviation (unbiased)
    sample_size = len(data)

    # Get Z-score for the given confidence level
    z_value = stats.norm.ppf((1 + confidence) / 2)

    # Compute Margin of Error (ME)
    margin_of_error = z_value * (sample_std / np.sqrt(sample_size))

    # Compute confidence interval
    lower_bound = sample_mean - margin_of_error
    upper_bound = sample_mean + margin_of_error

    return lower_bound, upper_bound, sample_mean

# Step 1: Generate Data from a Normal Distribution
np.random.seed(42)  # For reproducibility
data = np.random.normal(loc=50, scale=10, size=100)  # Mean=50, Std=10, Sample Size=100

# Step 2: Calculate Confidence Interval
ci_lower, ci_upper, sample_mean = confidence_interval(data)

# Step 3: Output Result
print(f"Sample Mean: {sample_mean:.2f}")
print(f"95% Confidence Interval: ({ci_lower:.2f}, {ci_upper:.2f})")
```

```python
# 24.
import numpy as np

def simulate_uniform_distribution(a, b, size=10000):
    """Simulate random variables from a continuous uniform distribution and calculate expected value."""
    # Simulate random variable from Uniform distribution
    random_variables = np.random.uniform(a, b, size)

    # Calculate the sample mean (which should approximate the expected value)
    sample_mean = np.mean(random_variables)

    # Calculate the theoretical expected value (E(X) = (a + b) / 2)
    expected_value = (a + b) / 2

    return sample_mean, expected_value

# Example usage
a, b = 0, 10  # Uniform distribution between 0 and 10
sample_mean, expected_value = simulate_uniform_distribution(a, b)

# Output the results
print(f"Sample Mean (Simulated): {sample_mean:.2f}")
print(f"Theoretical Expected Value: {expected_value:.2f}")
```

```python
# 26.
import numpy as np

def calculate_range_and_iqr(data):
    """Calculates the range and interquartile range (IQR) of a dataset."""

    # Calculate range
    data_range = np.max(data) - np.min(data)

    # Calculate interquartile range (IQR)
    Q1 = np.percentile(data, 25)
    Q3 = np.percentile(data, 75)
    IQR = Q3 - Q1

    return data_range, IQR

# Example: Generate data from a normal distribution
np.random.seed(42)  # For reproducibility
data = np.random.normal(loc=50, scale=10, size=1000)  # Mean=50, Std=10

# Calculate range and IQR
data_range, iqr = calculate_range_and_iqr(data)

# Output the results
print(f"Range of the dataset: {data_range:.2f}")
print(f"Interquartile Range (IQR) of the dataset: {iqr:.2f}")
```

```python
# 27.
import numpy as np
import matplotlib.pyplot as plt

def z_score_normalization(data):
    """Applies Z-score normalization to a dataset."""
    mean = np.mean(data)
    std_dev = np.std(data)
    normalized_data = (data - mean) / std_dev
    return normalized_data

def visualize_data_transformation(original_data, normalized_data):
    """Visualizes the original and normalized datasets."""
    plt.figure(figsize=(12, 6))

    # Plot original data
    plt.subplot(1, 2, 1)
    plt.hist(original_data, bins=30, color='blue', alpha=0.7, edgecolor='black')
    plt.title('Original Dataset')
    plt.xlabel('Value')
    plt.ylabel('Frequency')

    # Plot normalized data
    plt.subplot(1, 2, 2)
    plt.hist(normalized_data, bins=30, color='green', alpha=0.7, edgecolor='black')
    plt.title('Normalized Dataset (Z-score Normalization)')
    plt.xlabel('Value')
    plt.ylabel('Frequency')

    plt.tight_layout()
    plt.show()

# Generate data from a normal distribution
np.random.seed(42)
original_data = np.random.normal(loc=50, scale=10, size=1000)  # Mean=50, Std=10

# Apply Z-score normalization
normalized_data = z_score_normalization(original_data)

# Visualize the transformation
visualize_data_transformation(original_data, normalized_data)
```

Theoritical Answers

1. In probability theory, a random variable is a variable whose values are determined by the outcome of a random experiment. It is a function that maps the outcomes of a random process to numerical values.

  Random variables are often used to describe the results of uncertain events and are fundamental in statistical analysis.

2. There are two main types of random variables:

  **Discrete Random Variable:** Takes a finite or countably infinite set of distinct values (e.g., the number of heads in a series of coin flips).

  **Continuous Random Variable:** Takes an infinite number of values within a given range, typically represented by real numbers (e.g., the height of a person, the time it takes to complete a task).

3. The difference between discrete and continuous distributions lies in the type of random variable they describe:

  **Discrete Distribution:**

  Deals with discrete random variables that take distinct, countable values.

  Example: The number of heads in 10 coin flips.

  Probability is assigned to specific values.

  **Continuous Distribution:**

  Deals with continuous random variables that take an infinite number of values within a range.

  Example: The height of a person.

  Probability is assigned to intervals, not exact values (using probability density functions).

4. A Probability Distribution Function (PDF) is a function that describes the likelihood of a random variable taking on a particular value.

  The total area under a PDF curve (for continuous variables) equals 1, representing the certainty that the random variable will take some value within its range.
  

5. The key differences between a Cumulative Distribution Function (CDF) and a Probability Distribution Function (PDF) are:

  *  Definition:
  *   PDF (Probability Density Function): Describes the probability of a random variable taking a specific value (for discrete distributions) or the probability density at a point (for continuous distributions).
  *   CDF (Cumulative Distribution Function): Gives the probability that a random variable is less than or equal to a given value.

  *   Mathematical Relationship:
  *   The CDF is the integral (sum for discrete variables) of the PDF.
  *   The PDF is the derivative of the CDF (for continuous variables).

  *   Range:
  *   PDF values can be greater than 1 (for continuous distributions) but integrate to 1 over the entire range.
  *  CDF always ranges between 0 and 1 and is non-decreasing.

  *   Interpretation:
  *   PDF shows where values are most likely to occur.
  *   CDF shows the accumulated probability up to a certain value.







6. A discrete uniform distribution is a probability distribution where a finite number of outcomes are equally likely. Each outcome has the same probability.

  Example: Rolling a fair 6-sided die → Each face (1 to 6) has probability is 1/6.


7.

  Binary Outcomes:

  A Bernoulli distribution describes the probability of success or failure for a single trial (also called a Bernoulli trial).
  These outcomes are typically labeled as "success" (usually represented by 1) and "failure" (usually represented by 0).

  2. Single Parameter:

  It is characterized by a single parameter, often denoted as p, which represents the probability of success.
  The probability of failure is then 1 - p.
  
  3. Probability Mass Function (PMF):

  The PMF of a Bernoulli distribution is given by:
  P(X = 1) = p (probability of success)
  P(X = 0) = 1 - p (probability of failure)
  
  4. Mean and Variance:

  The mean (or expected value) of a Bernoulli distribution is p.
  The variance is p(1 - p).
  
  5. Applications:

  Bernoulli distributions are used to model events with binary outcomes, such as:
  Flipping a coin (heads or tails)
  Whether a customer will make a purchase (yes or no)
  Whether a machine is functioning (working or not working)

8.  The binomial distribution is a discrete probability distribution that describes the probability of getting exactly k successes in n independent Bernoulli trials (experiments with two possible outcomes: success or failure).

  P(X = k) = (n choose k) * p^k * (1 - p)^(n - k)

  The binomial distribution is used to model and analyze probabilities in various scenarios where you have a series of independent trials with two possible outcomes. Here are some examples:

  Quality Control: Determining the probability of finding a certain number of defective items in a batch.
  
  Medical Trials: Calculating the probability of a certain number of patients responding positively to a new drug.
  
  Marketing: Estimating the likelihood of a certain number of customers clicking on an advertisement.
  
  Genetics: Predicting the probability of inheriting a specific trait.
  
  Finance: Modeling the probability of a stock price going up or down a certain number of times.

9. The Poisson distribution is a discrete probability distribution that models the number of events occurring in a fixed interval of time or space, given that the events happen independently and at a constant average rate.

  P(X = k) = (e^-λ * λ^k) / k!

  where:

  e is the base of the natural logarithm (approximately 2.71828)
  
  λ is the average rate of events
  
  k is the number of events


  Applications:
  
    *   Counting the number of emails received per hour.
    *   Number of customer arrivals at a store per minute.
    *   Defects in manufacturing over a given length of material.
    *   Calls at a call center per hour.

22. Concept of Expected Value in a Probability Distribution:

  The expected value (EV) is the weighted average of all possible outcomes in a probability distribution, representing the long-term average if the experiment is repeated many times.

  Interpretation:

  The expected value is the center of mass of the probability distribution.
  
  It is the value you would "expect" to get on average if you repeated the experiment many times.
  
  It does not have to be a possible value of the random variable.

23. Relationship:

  A probability distribution describes the likelihood of different outcomes for a random variable. The expected outcome (or expected value) is a summary measure of this distribution, representing the average value you would expect to see if the random process were repeated many times.

  In essence:

  *   **Probability Distribution:** Provides the probabilities of individual outcomes.
  *   **Expected Outcome:** Is the weighted average of all possible outcomes, using their probabilities as weights.
